### histogram

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('07_opencv\\img\\img1.jpg', cv2.IMREAD_GRAYSCALE)
print(img.shape)
print(img.dtype)
print(img.min(), img.max())
```

```python
# cal histogram
# _, th_img = cv2.threshold(img, 200, 255, cv2.THRESH_BINARY)
# plt.imshow(th_img)
hst = cv2.calcHist([img], [0], None, [256], [0, 256])

plt.plot(hst)
plt.title('Grayscale Histogram')
plt.xlabel('Pixel Value')
plt.ylabel('Frequency')
plt.show()

```

```python
# 直方图均衡化
eq_img = cv2.equalizeHist(img)
eq_hst = cv2.calcHist([eq_img], [0], None, [256], [0, 256])
# plt.imshow(eq_img, cmap='gray')
plt.plot(eq_hst)
```

```python
# 彩色直方图
m_img = cv2.imread('07_opencv\\img\\smarties.png')
m1_hst = cv2.calcHist([m_img], [0], None, [256], [0, 256])
m2_hst = cv2.calcHist([m_img], [1], None, [256], [0, 256])
m3_hst = cv2.calcHist([m_img], [2], None, [256], [0, 256])

plt.plot(m1_hst)
```

```python
# 彩色直方图均衡化 (注意：COLOR_BGR2YCrCb，否则图像会失真)
ym_img = cv2.cvtColor(m_img, cv2.COLOR_BGR2YCrCb)
y, cr, cb = cv2.split(ym_img)
# b_img = cv2.equalizeHist(b)
# g_img = cv2.equalizeHist(g)
# r_img = cv2.equalizeHist(r)
# b_hst = cv2.calcHist([b_img], [0], None, [256], [0,256])
# plt.plot(b_hst)
# r.shape

yeq_img = cv2.equalizeHist(y)
cmb_img = cv2.merge((y, cr, cb))
cvt_img = cv2.cvtColor(cmb_img, cv2.COLOR_YCrCb2BGR)
# meq_img = cv2.merge([b_img, g_img, r_img])
plt.imshow(cvt_img)
```

```python
# 直方图比较
img1 = cv2.imread('07_opencv/img/img1.jpg', cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread('07_opencv/img/img2.jpg', cv2.IMREAD_GRAYSCALE)

img1_hst = cv2.calcHist([img1], [0], None, [256], [0, 256])
img2_hst = cv2.calcHist([img2], [0], None, [256], [0, 256])

sim = cv2.compareHist(img1_hst, img2_hst, cv2.HISTCMP_CORREL)
sim

```
