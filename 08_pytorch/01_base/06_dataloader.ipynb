## custom datasets

```python
from PIL import Image
import torch.utils.data as data
from torch.utils.data import DataLoader
import torchvision.transforms as transforms
import torch.utils.tensorboard as tb
import os


class SteelDataset(data.Dataset):
    def __init__(self, root, label, transforms=None):
        self.root = root
        self.label = label
        self.path = os.path.join(self.root, self.label)
        self.img_sets = os.listdir(self.path)
        self.transform = transforms

    def __getitem__(self, index):
        img_index = self.img_sets[index]
        img_path = os.path.join(self.path, img_index)
        label = self.label
        img = Image.open(img_path)

        if self.transform:
            img = self.transform(img)
        return label, img

    def __len__(self):
        return len(self.img_sets)
```

### create custom datasets

```python
root_dir = r'D:\Documents\DEV\Python\datasets\steel\casting_data\train'
ok_label = 'ok_front'
ng_label = 'def_front'

transform = transforms.Compose(
    [
        # transforms.Resize((300, 300)),
        transforms.ToTensor()
    ])
ok_dataset = SteelDataset(root_dir, ok_label, transforms=transform)
ng_dataset = SteelDataset(root_dir, ng_label, transforms=transform)

# merge dataset
trn_dataset = ok_dataset + ng_dataset
print(len(trn_dataset))
```

### add dataloader and tensorboard

```python
ok_loader = DataLoader(ok_dataset, batch_size=128, shuffle=True, drop_last=True)
ng_loader = DataLoader(ng_dataset, batch_size=128, shuffle=True)

writer = tb.SummaryWriter(log_dir='steel_summary')
step = 0
for label, img in ok_loader:
    writer.add_images('ok_img', img, step)
    step += 1
for label, img in ng_loader:
    writer.add_images('ng_img', img, step)
    step += 1
writer.close()
```
