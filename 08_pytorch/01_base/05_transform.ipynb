## transform

```python
from torchvision import transforms
from PIL import Image
import torch.utils.tensorboard as tb

img_path = r'D:\Documents\DEV\Python\datasets\steel\casting_data\train\ok_front\cast_ok_0_1.jpeg'
img = Image.open(img_path)
```

### to tensor

```python
trans_totensor = transforms.ToTensor()
img_tensor = trans_totensor(img)
print(img_tensor.dtype)
print(img_tensor.shape)  # ([C,H,W])
print(img_tensor.dim())  # 3
writer = tb.SummaryWriter('transform_tensor_board')
writer.add_image('tensor', img_tensor)
writer.close()
```

### Normalize

```python
print(img_tensor[0][0][0])
trans_norm = transforms.Normalize([0.5], [0.5])
img_norm_tensor = trans_norm(img_tensor)
print(img_norm_tensor[0][0][0])
writer.add_image('normalized_tensor', img_norm_tensor)
```

### resize

```python
print(img_tensor.shape)
trans_resize = transforms.Resize((600, 600))
img_tensor_resize = trans_resize(img_tensor)
print(img_tensor_resize.shape)
print(type(img_tensor_resize))
writer.add_image('resized_tensor', img_tensor_resize)
```

### compose

```python
# trans_comp = transforms.Compose([trans_totensor, trans_norm, trans_resize])     # 1
trans_comp = transforms.Compose(        # 2
    [
        transforms.ToTensor(),
        transforms.Normalize([0.5], [0.5]),
        transforms.Resize((600, 600))
    ])
img_tensor_comp = trans_comp(img)
print(img_tensor_comp.shape)
writer.add_image('tensor_comp', img_tensor_comp)
writer.close()
```
