## datetime

```python
from datetime import datetime

dt = datetime.now()
dt

# not permitted leading zero
dt = datetime(1999,5,23,18,1,1)
print(dt)

ts = dt.timestamp()
ts

t = 927453661.0
dt = datetime.fromtimestamp(t)
dt
```

## str to datetime

```python
from datetime import datetime

str = '1999,5,23 18:00:00'
trans = datetime.strptime(str, '%Y,%m,%d %H:%M:%S')
trans
```
