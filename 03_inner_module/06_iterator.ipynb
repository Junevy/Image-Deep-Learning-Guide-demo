## iterators

```python
import itertools

i = itertools.cycle('abc')
i
for ii in i:
    print(ii)
```

## repeat

```python
import itertools

re = itertools.repeat('ab', 3)

for r in re:
    print(re)
```

## iterators take while

```python
import itertools

c = itertools.count(1)

ns = itertools.takewhile(lambda x: x <= 10, c)
list(ns)
```

## chain

```python
import itertools

c = itertools.chain('xyz', 'abc', 'nmo')
for x in c :
    print(x)
```

## group by

```python
import itertools

# k: The name of some classifications
# g: Iterators grouped by k
for k, g in itertools.groupby('AABBBCCCC'):
    print(k, list(g))

for k, g in itertools.groupby("aaAAaBbCc", lambda c : c.upper()):
    print(k, list(g))
```
