## namedtuple -- such as struct

```python
from collections import namedtuple

point = namedtuple('point', ['x','y'])
point

p = point(1,2)  # generate a new namedtuple
p.x

isinstance(p, tuple)

t = (1,2,3)
t
isinstance(t, tuple)


circle = namedtuple('circle', ['x','y','radius'])
c = circle(1,2,3)
c
```

## deque -- the deque had exclusive methods that are addleft and popleft.

```python
from collections import deque

d = deque ([x for x in range(1,10)])
d

d.appendleft(33)
d

d.popleft()
d
```

## defaultdict -- The dict of possess a default value 

```python
from collections import defaultdict

dt = defaultdict(lambda: 'None')
dt

dt['key'] = 'value'
dt

n = dt['error']
n
```
