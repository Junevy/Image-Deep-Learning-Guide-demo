### frame extraction 

```python
import cv2 as cv
import matplotlib.pyplot as plt

cap = cv.VideoCapture('video/junevy.mp4')

# read one frame
ret, frame = cap.read()
plt.imshow(cv.cvtColor(frame, cv.COLOR_BGR2RGB))
```

```python
num = 0
step = 1

while True:
    ret, frame = cap.read()

    if not ret:
        break
    num += 1

    # if num % step == 0:
    cv.imwrite(f'/frame/{num}.jpg', frame)
    
```
