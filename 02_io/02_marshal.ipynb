## marshall / serialize 

```python
import pickle
from pathlib import Path

path = Path('02io_data/test.pkl')
path.parent.mkdir(exist_ok=True)

d = dict(name='juenvy', age=18, gender='male')

with path.open('wb') as f:
    pickle.dump(d, f)
```

## un marshall / deserialize

```python
with path.open('rb') as f:
    d = pickle.load(f)
    print(d)
```

## json serialize

```python
import json

d = dict(name='juenvy', age=18, gender='male')
j = json.dumps(d)

# unserialize
d = json.loads(j)
print(d)
```

## json improve

```python
# serialize object

import json

class student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
stu = student('junevy', 23)

def stu2json(stu):
    return {'name': stu.name, 'age':stu.age}

# print(json.dumps(stu, default=lambda obj: obj.__dict__))
sl = json.dumps(stu, default=stu2json)
print(sl)


# deserialize object
def dict2stu(dict):
    return student(dict['name'], dict['age'])

ss = json.loads(sl, object_hook=dict2stu)
print(ss.age)
```
