## string io write

```python
from io import StringIO

f = StringIO()
f.write('hello')
f.write(' ')
f.write('junevy')

f.getvalue()
```

## string read

```python
from io import StringIO

f = StringIO('hello\nhi\njunevy')

while True:
    s = f.readline()
    if s == '':
        break
    print(s)
```

## byte io

```python
from io import BytesIO

f = BytesIO()
f.write('english'.encode('utf-8'))
f.getvalue()
```

```python
from io import BytesIO

b = BytesIO(b'hello world')
b.getvalue()
print(b.getvalue())
```
