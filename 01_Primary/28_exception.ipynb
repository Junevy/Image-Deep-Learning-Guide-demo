```python
try:
    num = int(input('num:'))
    print(num + 3)
except TypeError as e:
    print('error:', e)
else:
    print('no error')
finally:
    print('finally')
```

```python
import logging

def foo():
    try:
        num = input('input:')
        int(num)
    except BaseException as e:
        logging.exception(e)

foo()
```

```python
import logging

def foo():
    try:
        num = input('input:')
        int(num)
    except BaseException as e:
        logging.exception(e)
        raise ValueError('input error!')

foo()
```
