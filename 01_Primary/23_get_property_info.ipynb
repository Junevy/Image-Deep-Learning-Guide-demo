```python
class test_attr:

    def __init__(self):
        self.x = 123

    def cal(self):
        return self.x * self.y
at = test_attr()

hasattr(at, 'x')
hasattr(at, 'y')    # false
#at.cal() # error

setattr(at, 'y', 20)
at.cal()

getattr(at, 'x')    # 123
#getattr(at, 's')    # error
getattr(at, 's', 404)   # 404
hasattr(at, 'cal')
```
